```python
# Explain print function with example
# The print() function in Python is used to output text or other data to the console. For example:
# print("Hello, World!") will display the text "Hello, World!" on the screen.
# It can also print multiple items separated by commas:
# print("Hello,", "World!") will output: Hello, World!
print("this", "is", "a", "test")  # This will output: this is a test
```

```python
print("this", "is", "a", "test", sep="-")  # This will output: this****is****a****test
```

```python
# Backslash Escape Sequences
print('This is a simple "Python World" script.')
print("This is a simple \"Python\" script.")
print("This is a simple \nPython script.")
print("This is a simple \\nPython script.")
print("This is a simple \tPython script.")
print("This is a simple \\tPython script.")
print("This is a simple \\tPython script.")
```
